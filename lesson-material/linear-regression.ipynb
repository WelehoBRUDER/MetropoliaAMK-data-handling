```python
import pandas as pd

df = pd.read_csv('../datasets/tvmarketing.csv')
df.head()
```

```python
import matplotlib.pyplot as plt

X = df.drop(columns='Sales') # inputs
y = df['Sales'] # targets
plt.scatter(X, y)
plt.xlabel('TV Advertising (1000$)')
plt.ylabel('Sales revenue (M$)')
plt.show()
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X, y)
```

```python
b0 = model.intercept_
b1 = model.coef_[0]
print(f'Intercept: {b0:.2f}, Coefficient: {b1:.4f}')
```

```python
import numpy as np

xs = np.linspace(X.min(), X.max())
ys = b0 + b1 * xs
plt.plot(xs, ys)
plt.scatter(X, y)
plt.xlabel('TV Advertising (1000$)')
plt.ylabel('Sales revenue (M$)')
plt.show()
```
